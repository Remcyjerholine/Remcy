```python
import random

# 🎰 Configuration
FRUITS = ["🍒", "🍋", "🍊", "🍉", "⭐"]  # cherries, lemon, orange, watermelon, star
REELS = 3
SYMBOL_COUNT = {
    "🍒": 2,
    "🍋": 3,
    "🍊": 4,
    "🍉": 3,
    "⭐": 1
}
SYMBOL_VALUE = {
    "🍒": 5,
    "🍋": 4,
    "🍊": 3,
    "🍉": 2,
    "⭐": 10
}

def spin():
    """Generate one spin: a list of symbols, one per reel."""
    all_symbols = []
    for sym, count in SYMBOL_COUNT.items():
        all_symbols += [sym] * count
    return [random.choice(all_symbols) for _ in range(REELS)]

def get_payoutt(symbols, bet):
    """
    Calculate payout multiplier based on matching fruits.
    - 3 of kind: value × bet
    - 2 of a kind: half value × bet
    - else: lose bet
    """
    if symbols.count(symbols[0]) == REELS:
        # All three match
        return SYMBOL_VALUE[symbols[0]] * bet
    elif len(set(symbols)) == REELS - 1:
        # Exactly two match
        # Find the symbol that appears twice
        for s in SYMBOL_VALUE:
            if symbols.count(s) == 2:
                return (SYMBOL_VALUE[s] // 2) * bet
    return -bet

def print_spin(symbols):
    print(" | ".join(symbols))

def play_slot_machine():
    balance = 100
    while balance > 0:
        print(f"\nBalance: ₹{balance}")
        bet = input("Enter bet amount (or 'q' to quit): ")
        if bet.lower() == "q":
            break
        if not bet.isdigit() or int(bet) <= 0 or int(bet) > balance:
            print("Invalid bet.")
            continue
        bet = int(bet)
        result = spin()
        print_spin(result)
        payout = get_payoutt(result, bet)
        balance += payout
        if payout > 0:
            print(f"You won ₹{payout}!")
        else:
            print(f"You lost ₹{bet}.")
    print(f"Game over. Final balance: ₹{balance}")

if __name__ == "__main__":
    play_slot_machine()
```

```python
#inherritece
class Parent:   #parent class is the parent 
    def show(self):
        print("I am the parent")   #the function inside the class parent

class Child(Parent):   #the inherited class under the parent class 
    pass                            # no funtion under the child class it takes the function inder the parent class as its inherited from it

c = Child()     #creating a instance under child class
c.show()  #printing the instance ...it print the funtion from parent class
```

```python
#encapsulation
class student:
    def __init__(self):
        self.name="jerholoine"               #an public variable ...starts with the variable itsselff (can be accesed outside the class)
        self.__age=18                        # the private variable   ...starts with two underscore(cannot)
    def get_age(self):                       #getter -to get the output and to access the variable outside the class
        return self.__age
    def set_age(self,age)    :                 #setter -used to update or set a value to the variable
        self.__age=age
s=student()        
print(s.name)
print(s.get_age())
s.set_age(20)               #assining the valueto the variable set
print(s.get_age())
```

```python
#polymorphism
class dog:                #dog is sthe one class with method "sound"
    def sound(self):
        print("the dog barks")
class bird:             #bird is another class with same methos sound
    def sound(self):
        print("the bird chips")
for animal in (dog(),bird()):          #but both class has same methods ...so its polymorphism
    animal.sound()
```

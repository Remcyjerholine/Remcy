```python
#PYTHON COMPUND INTREST
principle=0
rate=0
time=0
while True:
    principle=float(input("enter the value of the principle:"))
    if principle<=0:
        print("principle cant be less than 0 or 0")
    else :    
        break
while True:
    rate=float(input("enter the value of the rate:"))
    if rate<=0:
        print("rate cant be less than 0 or 0")
    else :    
        break
while True:
    time=int(input("enter the value of the time:"))
    if time<=0:
        print("time cant be less than 0 or 0")
    else :    
        break
compound_intrest=principle*pow((1+rate/100),time)
print(f"the compund intrest is: {compound_intrest:.2f}")
                        
    
```

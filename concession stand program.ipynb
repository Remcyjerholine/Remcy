```python
#CONCESSION STAND PROGRAM
menu={"cacins milkshake":40,"kinder joy":35,"bingo":10,"brownie":25,"buttermilk":10,"lays":30,"cake":15}
cart=[]
total=0
print("-------mennu------")
for key,value in menu.items():
    print(f"{key:.10}:${value:.2f}")
print("--------------------")
while True:
    food=input("select the item from the menu(q to quit):").lower()
    if food=='q':
        break
    elif menu.get(food) is not None:
        cart.append(food)
print("--------your order--------")        
for food in cart:
    total+=menu.get(food)
    print(food,end=" ")
print()
print(f"total is:${total:.2f}")
```

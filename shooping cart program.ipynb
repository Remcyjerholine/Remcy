```python
foods=[]
prices=[]
total=0
while True:
    food=input("enter the food to buy(q or Q to qiut):")
    if food.lower()=="q":
        break
    else:
        price=float(input("enter the price of the {food}:$"))
        foods.append(food)
        prices.append(price)
print("----your cart----")        
for food in foods:
    print(food,end=" ")
for price in prices:
    total +=price
print()
print(f"your total is:{total}")
```

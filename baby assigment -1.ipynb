```python
bmi=int(input("enter the BMI index:"))
if(bmi<18.5):
    print("underweight")
elif(18.5<=bmi<=24.9):
    print("normal")
elif(25<=bmi<=29.9)    :
    print("overweight")
elif(30<=bmi<=34.9):
     print("obesity (mild--class1)")
elif(35<=bmi<=39.9)           :
    print("obessity(moderate--class2)")
else:
    print("obesity(morbid--class3)")
```
